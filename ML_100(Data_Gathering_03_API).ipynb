#Loop to get all pages data

```python
df = pd.DataFrame()
df
```

```python
dfs = []
for i in range(1, 1002):
  response = requests.get(f"https://api.themoviedb.org/3/discover/movie?api_key=d991f869167828b80b9f0bb07c3d8911&page={i}")
  if response.status_code == 200:
    results = response.json().get('results', [])
    if not results:
      break
    temp_df = pd.DataFrame(results).head(4)[['id', 'title', 'adult', 'video', 'vote_count', 'vote_average']]
    dfs.append(temp_df)
  else:
    break

if dfs:
  df = pd.concat(dfs, ignore_index=True)
else:
  df = pd.DataFrame()
```
